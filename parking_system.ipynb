```python
# Smart Parking System

parking_slots = {
    1: None,
    2: None,
    3: None,
    4: None,
    5: None
}


def show_slots():
    print("\n===== PARKING SLOTS =====")

    for slot, vehicle in parking_slots.items():

        if vehicle is None:
            print("Slot", slot, ": Available")
        else:
            print("Slot", slot, ":", vehicle)


def park_vehicle():
    vehicle = input("\nEnter Vehicle Number: ").upper()

    # Check whether vehicle is already parked
    for slot, parked_vehicle in parking_slots.items():
        if parked_vehicle == vehicle:
            print("Vehicle is already parked!")
            return

    # Find an empty slot
    for slot, parked_vehicle in parking_slots.items():

        if parked_vehicle is None:
            parking_slots[slot] = vehicle

            print("Vehicle parked successfully!")
            print("Your Parking Slot:", slot)
            return

    print("Sorry! Parking is full.")


def remove_vehicle():
    vehicle = input("\nEnter Vehicle Number: ").upper()

    for slot, parked_vehicle in parking_slots.items():

        if parked_vehicle == vehicle:
            parking_slots[slot] = None

            print("Vehicle removed successfully!")
            print("Released Slot:", slot)
            return

    print("Vehicle not found!")


def search_vehicle():
    vehicle = input("\nEnter Vehicle Number: ").upper()

    for slot, parked_vehicle in parking_slots.items():

        if parked_vehicle == vehicle:
            print("Vehicle Found!")
            print("Vehicle Number:", vehicle)
            print("Parking Slot:", slot)
            return

    print("Vehicle not found!")


while True:

    print("\n==============================")
    print("      SMART PARKING SYSTEM")
    print("==============================")

    print("1. Show Parking Slots")
    print("2. Park Vehicle")
    print("3. Remove Vehicle")
    print("4. Search Vehicle")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        show_slots()

    elif choice == "2":
        park_vehicle()

    elif choice == "3":
        remove_vehicle()

    elif choice == "4":
        search_vehicle()

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
